```python
import altair as alt
import polars as pl
import numpy as np

df = pl.DataFrame(
    {
        # "x": [0, 1, 2, 3, 3, 4, 5],
        "y": [0, 1, 2, None, 3, 4, 5],
    }
)
df.to_pandas()
```

```python
(
    alt.Chart(df.to_pandas())
    .transform_calculate(
        metric=f'datum.y'
    )
    .transform_filter("isValid(datum.metric)")
    .transform_window(
        cdf="cume_dist()", 
        sort=[{"field": "x"}],
    )
    .encode(x='metric:Q', y='cdf:Q')
    .mark_line(
        point=True,             
        interpolate="step-after",
    )
)
```
